# GTFS expected departure_time analysis

This notebook will explore the analysis obtained in the stop_times.txt file from the GTFS specification. It is a subgroup of that file with a demand of line 401 on a LA (labour) day.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


data_frame = pd.read_csv("departures.csv", delimiter=",")

data_frame.head(n=10)
```

Now, we will obtain only the stop_idx == 1, this is the headway for each direction. We will add a column with a basic comparaison to add direction 1 for 2646 code and direction 2 for 5727 code.

```python
def time_to_u32 (time:str) -> int :
    h,m,s, i = 0

    accum = ""
    for element in time:
        if element == ":":
            if i == 0:
                h = int(accum)
                accum = ""
            elif i == 2:
                m = int(accum)
                accum = ""
            else:
                s = int(accum)
                accum = ""

        else:
            accum += element

    return h*3600 + m*60 + s




data_headways = data_frame[data_frame["stop_idx"] == 1]

data_headways["Direction"] = data_headways["stop_code"].apply(lambda x: 1 if x == 2646 else 2)




print(f"My calcs, 191 rows, can match for a single day??? Thats far beyond the 163 mean departures")
```
