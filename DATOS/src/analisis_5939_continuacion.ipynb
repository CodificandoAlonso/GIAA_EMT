```python
from fussion_gauss import *
import ast




new_dataframe = new_dataframe.iloc[1:-1]


```

```python


#stablish threshold
threshold = 0.97

dictionary_final = {}


for iteration, row in new_dataframe.iterrows():






    stopping_condition = False


    dictionary_per_slot = {}

    for timeslot in range (5, 24):
        list = []
        list.append(row[f"Mean_{timeslot}"])
        list.append(row[f"std_{timeslot}"])
        list.append(row[f"IsNormal_{timeslot}"])
        list.append(row[f"outliers_{timeslot}"])
        list.append(row[f"instances_{timeslot}"])
        dictionary_per_slot[timeslot] = list


    elements_fused = 0

    while stopping_condition == False:

        #calc the pair by pair bhattacharyya distance and coeff

        elements = [key for key in dictionary_per_slot.keys()]
        distances = {}
        iteration = 0
        for element_to_calc in elements:


            #Add only its calculus with the following element
            if element_to_calc == elements[0]:

                either_element = elements[1]
                bhatt_distance = bhattacharyya_distance(dictionary_per_slot[element_to_calc][0], dictionary_per_slot[either_element][0], dictionary_per_slot[element_to_calc][1], dictionary_per_slot[either_element][1])

                bhatt_coeff = bhattacharyya_coefficient(dictionary_per_slot[element_to_calc][0], dictionary_per_slot[either_element][0], dictionary_per_slot[element_to_calc][1], dictionary_per_slot[either_element][1])

                tuple_to_add = (element_to_calc, either_element)
                distances[tuple_to_add] = (bhatt_distance, bhatt_coeff)
                iteration += 1

            elif element_to_calc == elements[len(elements)-1]:
                pass

            else:
                    either_element = elements[iteration +1]
                    print(f"Calculating bhattacharyya dist with {dictionary_per_slot[element_to_calc][0]}, {dictionary_per_slot[element_to_calc][1]}, {dictionary_per_slot[either_element][0]}, {dictionary_per_slot[either_element][1]}")
                    bhatt_distance = bhattacharyya_distance(dictionary_per_slot[element_to_calc][0], dictionary_per_slot[either_element][0], dictionary_per_slot[element_to_calc][1], dictionary_per_slot[either_element][1])

                    bhatt_coeff = bhattacharyya_coefficient(dictionary_per_slot[element_to_calc][0], dictionary_per_slot[either_element][0], dictionary_per_slot[element_to_calc][1], dictionary_per_slot[either_element][1])

                    tuple_to_add = (element_to_calc, either_element)
                    distances[tuple_to_add] = (bhatt_distance, bhatt_coeff)
                    iteration +=1



        minimum = 999
        key_elem = []
        value_dist = []
        for key, value in distances.items():
            if distances[key][0] < minimum:
                minimum = distances[key][0]
                key_elem = key
                value_dist = value



        print(f"Con key {key_elem} hay este val {value_dist}")

        #Check if COEFF is above threshold
        if value_dist[1] < threshold:
            stopping_condition = True
            continue

        #FUSSION THE 2 GAUSS DISTROS

        elements_1 = dictionary_per_slot[key_elem[0]][-1]
        elements_2 = dictionary_per_slot[key_elem[1]][-1]

        elements_1 = ast.literal_eval(elements_1)
        elements_2 = ast.literal_eval(elements_2)


        gauss_hypothesis, instances_involved, mean, std  = is_normal_simplified(elements_1 + elements_2)
        print(f" Gauss new obtained with {gauss_hypothesis}, {mean}, {std}")

        varname = f"FUSS_{key_elem[0]}-{key_elem[1]}"


        i = 0
        for element in elements:
            if element == key_elem[0]:
                elements[i] = varname
                elements.pop(i+1)
            i += 1



        new_dictionary = {}

        for new_key in elements:
            if new_key in dictionary_per_slot.keys():
                old_info = dictionary_per_slot[new_key]
            else:
                old_info = [mean, std, gauss_hypothesis, str(instances_involved)]
            new_dictionary[new_key] = old_info


        dictionary_per_slot = new_dictionary


    dictionary_final[row["Stop"]] = dictionary_per_slot



for key, value in dictionary_final.items():
    print(f"For stop {key}, we obtain this distribution")
    print(f"\tWe obtain the following timeslots:")
    for key2, val2 in value.items():
        print(f"\t\tFor timeslot {key2}, we obtain this distribution {val2[0]} of mean and {val2[1]} of std and hyp of {val2[2]}")
```
